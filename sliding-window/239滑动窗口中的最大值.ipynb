```python

import heapq
class Solution(object):
    def maxSlidingWindow(self, nums, k):
        n = len(nums)
        # 注意 Python 默认的优先队列是小根堆
        q = [(-nums[i], i) for i in range(k)]
        heapq.heapify(q)

        ans = [-q[0][0]]
        for i in range(k, n):
            left=i-k
            heapq.heappush(q, (-nums[i], i))
            while q[0][1] <= left:
                heapq.heappop(q)
            ans.append(-q[0][0])
        return ans


solution=Solution()
solution.maxSlidingWindow([1,3,-1,-3,5,3,6,7], k = 3)
```
