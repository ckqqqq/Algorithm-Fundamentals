# 链表_21合并两个有序链表

> 自动合成题干：文件无题面内容，需自行补充。


```python
class node:
    def __init__(self,next=None,val=0):
        self.next=next
        self.val=val
def solution(head1,head2):
    empty_head=node()
    cur=empty_head
    while head1 and head2:
        if head1.val<head2.val:
            cur.next=head1
            cur=cur.next
            head1=head1.next
        else:
            cur.next=head2
            cur=cur.next
            head2=head2.next
    if head1==None:
        cur.next=head2
    else:#head2==None
        cur.next=head1
    return empty_head.next
```
