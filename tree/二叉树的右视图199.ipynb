# 二叉树的右视图199

> 自动合成题干：文件无题面内容，需自行补充。


```python
class Solution(object):
    dep_res={}
    dep_max=-1
    def rightSideView(self, root):
        def dfs_(r,d):
            if r==None:
                return 
            dfs_(r.left,d+1)
            self.dep_res[d]=r.val
            self.dep_max=max(self.dep_max,d)
            dfs_(r.right,d+1)
        dfs_(root,0)
        res=[self.dep_res[d] for d in range(0,self.dep_max+1)]
        return res
```
