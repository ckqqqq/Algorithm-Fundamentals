```python
#统计是不是2的次幂
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        return n>0 and n&n-1
        
```

```python
#0-n+1统计变量名
class Solution:
    def countBits(self, n: int) -> List[int]:
        ans=[]
        for i in range(n+1):
            count=0
            while n!=0:#这里写成了n，有问题，注意复制黏贴的时候小心变量名
                n=n&n-1#去除最后一个一
                #n&-n 获取最后一个一
                count+=1
            ans.append(count)
        return ans

       
```
