* GSPO：解决 MoE 训练中 GRPO 不稳定的问题。
    * 专家激活波动性是关键问题。因为新旧策略可能激活不同的专家，带来结构性偏差
    * Routing Replay 会记录 推理时的路由激活，并在训练时强制使用相同激活路径
    * GSPO 的算法希望抛弃掉 GRPO 的 token-level objective，而是把 importance rate 直接用在 sequence-level 上，这也就自然得引入了 GSPO 的优化算法目标，即把 token-level 的 importance rate 换成了 sequence-level 的 importance rate。
